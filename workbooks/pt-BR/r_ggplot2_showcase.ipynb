# Vitrine do ggplot2

Gráficos interativos com ggplot2 utilizando o tema escuro do SciREPL — aplicado automaticamente, sem necessidade de configuração adicional.

Utiliza conjuntos de dados integrados do R (`mpg`, `mtcars`). Nenhum download de dados é necessário.

```r
install.packages("ggplot2")
library(ggplot2)
cat("ggplot2 carregado — theme_scirepl() aplicado automaticamente\n")
```

## 1. Gráfico de Dispersão com Linha de Tendência

```r
ggplot(mpg, aes(displ, hwy, color = class)) +
  geom_point(size = 2, alpha = 0.7) +
  geom_smooth(method = "lm", se = FALSE, color = "#58a6ff") +
  labs(title = "Tamanho do motor vs MPG na estrada",
       x = "Cilindrada (L)", y = "MPG na estrada")
```

## 2. Gráfico de Barras em Painéis (Faceted)

```r
ggplot(mpg, aes(class, fill = drv)) +
  geom_bar(position = "dodge") +
  facet_wrap(~year) +
  labs(title = "Classes de veículos por tipo de tração (1999 vs 2008)",
       x = "Classe", y = "Contagem") +
  theme(axis.text.x = element_text(angle = 45, hjust = 1))
```

## 3. Gráfico de Densidade

```r
ggplot(mpg, aes(hwy, fill = drv)) +
  geom_density(alpha = 0.5) +
  labs(title = "Distribuição de MPG na estrada por tipo de tração",
       x = "MPG na estrada", y = "Densidade")
```

## 4. Boxplot com Dispersão de Pontos (Jitter)

```r
ggplot(mpg, aes(class, hwy, fill = class)) +
  geom_boxplot(alpha = 0.7, show.legend = FALSE) +
  geom_jitter(width = 0.2, alpha = 0.3, size = 1) +
  labs(title = "MPG na estrada por classe de veículo",
       x = "Classe", y = "MPG na estrada") +
  coord_flip()
```

## 5. Mapa de Calor de Correlação

```r
corr <- round(cor(mtcars), 2)
vars <- colnames(mtcars)
grid <- expand.grid(x = vars, y = vars)
grid$value <- as.vector(corr)
ggplot(grid, aes(x, y, fill = value)) +
  geom_tile(color = "#30363d") +
  geom_text(aes(label = ifelse(abs(value) > 0.5, value, "")),
            color = "#e6edf3", size = 2.5) +
  scale_fill_gradient2(low = "#f85149", mid = "#161b22", high = "#58a6ff",
                       midpoint = 0, limits = c(-1, 1)) +
  labs(title = "Mapa de calor de correlação mtcars", fill = "r") +
  theme(axis.text.x = element_text(angle = 45, hjust = 1))
```

## Resumo

Todos os gráficos acima utilizam `theme_scirepl()`, aplicado automaticamente quando o ggplot2 é carregado no SciREPL.

Personalize qualquer gráfico com as sobreposições padrão de `+ theme(...)` do ggplot2.
