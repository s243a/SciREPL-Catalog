# ggplot2 शोकेस

SciREPL की डार्क थीम के साथ इंटरैक्टिव ggplot2 चार्ट — स्वचालित रूप से लागू, किसी सेटअप की आवश्यकता नहीं।

R के अंतर्निहित डेटासेट (`mpg`, `mtcars`) का उपयोग करता है। कोई डेटा डाउनलोड करने की आवश्यकता नहीं है।

```r
install.packages("ggplot2")
library(ggplot2)
cat("ggplot2 लोड हुआ — theme_scirepl() स्वचालित रूप से लागू किया गया\n")
```

## 1. ट्रेंड लाइन के साथ स्कैटर प्लॉट

```r
ggplot(mpg, aes(displ, hwy, color = class)) +
  geom_point(size = 2, alpha = 0.7) +
  geom_smooth(method = "lm", se = FALSE, color = "#58a6ff") +
  labs(title = "इंजन का आकार बनाम हाईवे MPG",
       x = "विस्थापन (L)", y = "हाईवे MPG")
```

## 2. पहलूयुक्त (Faceted) बार चार्ट

```r
ggplot(mpg, aes(class, fill = drv)) +
  geom_bar(position = "dodge") +
  facet_wrap(~year) +
  labs(title = "ड्राइव प्रकार के अनुसार वाहन श्रेणियां (1999 बनाम 2008)",
       x = "वाहन श्रेणी", y = "संख्या") +
  theme(axis.text.x = element_text(angle = 45, hjust = 1))
```

## 3. घनत्व (Density) प्लॉट

```r
ggplot(mpg, aes(hwy, fill = drv)) +
  geom_density(alpha = 0.5) +
  labs(title = "ड्राइव प्रकार के अनुसार हाईवे MPG का वितरण",
       x = "हाईवे MPG", y = "घनत्व")
```

## 4. जिटर के साथ बॉक्सप्लॉट (Boxplot with Jitter)

```r
ggplot(mpg, aes(class, hwy, fill = class)) +
  geom_boxplot(alpha = 0.7, show.legend = FALSE) +
  geom_jitter(width = 0.2, alpha = 0.3, size = 1) +
  labs(title = "वाहन श्रेणी के अनुसार हाईवे MPG",
       x = "वाहन श्रेणी", y = "हाईवे MPG") +
  coord_flip()
```

## 5. सहसंबंध हीटमैप

```r
corr <- round(cor(mtcars), 2)
vars <- colnames(mtcars)
grid <- expand.grid(x = vars, y = vars)
grid$value <- as.vector(corr)
ggplot(grid, aes(x, y, fill = value)) +
  geom_tile(color = "#30363d") +
  geom_text(aes(label = ifelse(abs(value) > 0.5, value, "")),
            color = "#e6edf3", size = 2.5) +
  scale_fill_gradient2(low = "#f85149", mid = "#161b22", high = "#58a6ff",
                       midpoint = 0, limits = c(-1, 1)) +
  labs(title = "mtcars सहसंबंध हीटमैप", fill = "r") +
  theme(axis.text.x = element_text(angle = 45, hjust = 1))
```

## सारांश

उपरोक्त सभी प्लॉट `theme_scirepl()` का उपयोग करते हैं, जो SciREPL में ggplot2 लोड होने पर स्वचालित रूप से लागू होता है।

मानक ggplot2 `+ theme(...)` ओवरराइड के साथ किसी भी प्लॉट को अनुकूलित करें।
