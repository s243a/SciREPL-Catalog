# Statistik dengan R

Pengujian hipotesis, regresi, dan interval kepercayaan menggunakan kumpulan data bawaan R.

Tidak perlu mengunduh data atau memasang paket — hanya menggunakan R dasar (Base R).

## 1. Statistik Deskriptif

```r
data(mtcars)
cat("Kumpulan data: mtcars (", nrow(mtcars), " mobil, ", ncol(mtcars), " variabel)\n\n")
summary(mtcars[, c("mpg", "hp", "wt", "disp")])
```

## 2. Uji-t Dua Sampel

Apakah mobil transmisi manual memiliki efisiensi bahan bakar (MPG) yang lebih baik daripada otomatis?

```r
auto <- mtcars$mpg[mtcars$am == 0]
manual <- mtcars$mpg[mtcars$am == 1]

cat("Otomatis:", round(mean(auto), 1), "mil per galon (n =", length(auto), ")\n")
cat("Transmisi manual:   ", round(mean(manual), 1), "mil per galon (n =", length(manual), ")\n\n")

t_result <- t.test(manual, auto, alternative = "greater")
print(t_result)

cat("\nKesimpulan:",
    ifelse(t_result$p.value < 0.05,
           "Tolak H0 — mobil manual memiliki MPG yang signifikan lebih tinggi",
           "Gagal menolak H0"))
```

## 3. Uji Chi-Kuadrat

Apakah jumlah silinder dan jenis transmisi saling independen?

```r
tab <- table(Cylinders = mtcars$cyl, Transmission = mtcars$am)
colnames(tab) <- c("Otomatis", "Manual")
print(tab)
cat("\n")
chisq.test(tab)
```

## 4. Regresi Linier Berganda

```r
model <- lm(mpg ~ wt + hp + am, data = mtcars)
summary(model)
```

## 5. Diagnostik Regresi

```r
par(mfrow = c(2, 2))
plot(model)
```

## 6. Interval Kepercayaan

```r
ci <- confint(model, level = 0.95)
cat("Interval Kepercayaan 95%:\n")
print(round(ci, 4))
```

```r
coefs <- coef(model)[-1]
ci_vals <- ci[-1, ]
n <- length(coefs)

par(mfrow = c(1, 1), mar = c(5, 8, 4, 2))
plot(coefs, 1:n, xlim = range(ci_vals),
     pch = 19, col = "#58a6ff", cex = 1.5,
     yaxt = "n", xlab = "Estimasi", ylab = "",
     main = "Interval Kepercayaan 95% untuk Koefisien")
axis(2, at = 1:n, labels = names(coefs), las = 1)
segments(ci_vals[, 1], 1:n, ci_vals[, 2], 1:n,
         lwd = 3, col = "#58a6ff")
abline(v = 0, lty = 2, col = "#f85149", lwd = 1.5)
```

## 7. ANOVA Satu Arah (One-Way ANOVA)

Apakah efisiensi bahan bakar (MPG) berbeda secara signifikan antar jumlah silinder?

```r
anova_model <- aov(mpg ~ factor(cyl), data = mtcars)
summary(anova_model)
cat("\nPerbandingan Post-hoc Tukey HSD:\n")
TukeyHSD(anova_model)
```

```r
boxplot(mpg ~ cyl, data = mtcars,
        main = "MPG Berdasarkan Jumlah Silinder",
        xlab = "Silinder", ylab = "Mil per Galon",
        col = c("#58a6ff", "#a371f7", "#f85149"))
```

## Ringkasan

- **Uji-t Welch**: Dalam perbandingan satu sisi yang belum disesuaikan, mobil manual memiliki rata-rata MPG lebih tinggi
- **Chi-kuadrat**: Tabel kontingensi menunjukkan adanya hubungan tetapi jumlah ekspektasi yang rendah memicu peringatan perkiraan, jadi tafsirkan hasil ini dengan hati-hati
- **Regresi**: Berat dan tenaga kuda (horsepower) adalah prediktor negatif yang signifikan setelah penyesuaian; jenis transmisi tidak signifikan dalam model ini
- **ANOVA**: MPG berbeda secara signifikan antara kelompok 4, 6, dan 8 silinder; hasil Tukey mengidentifikasi perbedaan berpasangan
