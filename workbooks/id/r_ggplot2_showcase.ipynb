# Pameran ggplot2

Bagan interaktif ggplot2 dengan tema gelap SciREPL — diterapkan secara otomatis, tidak memerlukan pengaturan.

Menggunakan kumpulan data bawaan R (`mpg`, `mtcars`). Tidak perlu mengunduh data.

```r
install.packages("ggplot2")
library(ggplot2)
cat("ggplot2 dimuat — theme_scirepl() diterapkan otomatis\n")
```

## 1. Plot Sebar (Scatter Plot) dengan Garis Tren

```r
ggplot(mpg, aes(displ, hwy, color = class)) +
  geom_point(size = 2, alpha = 0.7) +
  geom_smooth(method = "lm", se = FALSE, color = "#58a6ff") +
  labs(title = "Ukuran Mesin vs MPG Jalan Raya",
       x = "Kapasitas Mesin (L)", y = "MPG Jalan Raya")
```

## 2. Bagan Batang Berpanel (Faceted)

```r
ggplot(mpg, aes(class, fill = drv)) +
  geom_bar(position = "dodge") +
  facet_wrap(~year) +
  labs(title = "Kelas Kendaraan berdasarkan Tipe Penggerak (1999 vs 2008)",
       x = "Kelas", y = "Jumlah") +
  theme(axis.text.x = element_text(angle = 45, hjust = 1))
```

## 3. Plot Densitas

```r
ggplot(mpg, aes(hwy, fill = drv)) +
  geom_density(alpha = 0.5) +
  labs(title = "Distribusi MPG Jalan Raya berdasarkan Tipe Penggerak",
       x = "MPG Jalan Raya", y = "Densitas")
```

## 4. Boxplot dengan Jitter

```r
ggplot(mpg, aes(class, hwy, fill = class)) +
  geom_boxplot(alpha = 0.7, show.legend = FALSE) +
  geom_jitter(width = 0.2, alpha = 0.3, size = 1) +
  labs(title = "MPG Jalan Raya berdasarkan Kelas Kendaraan",
       x = "Kelas", y = "MPG Jalan Raya") +
  coord_flip()
```

## 5. Heatmap Korelasi

```r
corr <- round(cor(mtcars), 2)
vars <- colnames(mtcars)
grid <- expand.grid(x = vars, y = vars)
grid$value <- as.vector(corr)
ggplot(grid, aes(x, y, fill = value)) +
  geom_tile(color = "#30363d") +
  geom_text(aes(label = ifelse(abs(value) > 0.5, value, "")),
            color = "#e6edf3", size = 2.5) +
  scale_fill_gradient2(low = "#f85149", mid = "#161b22", high = "#58a6ff",
                       midpoint = 0, limits = c(-1, 1)) +
  labs(title = "Heatmap Korelasi mtcars", fill = "r") +
  theme(axis.text.x = element_text(angle = 45, hjust = 1))
```

## Ringkasan

Semua plot di atas menggunakan `theme_scirepl()`, yang diterapkan secara otomatis saat ggplot2 dimuat di SciREPL.

Sesuaikan plot apa pun dengan penggantian `+ theme(...)` standar ggplot2.
