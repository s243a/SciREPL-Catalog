# Vitrine ggplot2

Graphiques interactifs avec ggplot2 utilisant le thème sombre de SciREPL — appliqué automatiquement, sans configuration requise.

Utilise les jeux de données intégrés de R (`mpg`, `mtcars`). Aucun téléchargement de données n'est requis.

```r
install.packages("ggplot2")
library(ggplot2)
cat("ggplot2 chargé — theme_scirepl() appliqué automatiquement\n")
```

## 1. Nuage de points avec courbe de tendance

```r
ggplot(mpg, aes(displ, hwy, color = class)) +
  geom_point(size = 2, alpha = 0.7) +
  geom_smooth(method = "lm", se = FALSE, color = "#58a6ff") +
  labs(title = "Cylindrée vs MPG sur autoroute",
       x = "Cylindrée (L)", y = "MPG sur autoroute")
```

## 2. Diagramme en barres à facettes (Faceted)

```r
ggplot(mpg, aes(class, fill = drv)) +
  geom_bar(position = "dodge") +
  facet_wrap(~year) +
  labs(title = "Classes de véhicules par type de transmission (1999 vs 2008)",
       x = "Classe", y = "Nombre") +
  theme(axis.text.x = element_text(angle = 45, hjust = 1))
```

## 3. Graphique de densité

```r
ggplot(mpg, aes(hwy, fill = drv)) +
  geom_density(alpha = 0.5) +
  labs(title = "Distribution du MPG sur autoroute par type de transmission",
       x = "MPG sur autoroute", y = "Densité")
```

## 4. Boîte à moustaches avec dispersion de points (Jitter)

```r
ggplot(mpg, aes(class, hwy, fill = class)) +
  geom_boxplot(alpha = 0.7, show.legend = FALSE) +
  geom_jitter(width = 0.2, alpha = 0.3, size = 1) +
  labs(title = "MPG sur autoroute par classe de véhicule",
       x = "Classe", y = "MPG sur autoroute") +
  coord_flip()
```

## 5. Carte thermique de corrélation

```r
corr <- round(cor(mtcars), 2)
vars <- colnames(mtcars)
grid <- expand.grid(x = vars, y = vars)
grid$value <- as.vector(corr)
ggplot(grid, aes(x, y, fill = value)) +
  geom_tile(color = "#30363d") +
  geom_text(aes(label = ifelse(abs(value) > 0.5, value, "")),
            color = "#e6edf3", size = 2.5) +
  scale_fill_gradient2(low = "#f85149", mid = "#161b22", high = "#58a6ff",
                       midpoint = 0, limits = c(-1, 1)) +
  labs(title = "Carte thermique de corrélation mtcars", fill = "r") +
  theme(axis.text.x = element_text(angle = 45, hjust = 1))
```

## Résumé

Tous les graphiques ci-dessus utilisent `theme_scirepl()`, appliqué automatiquement lors du chargement de ggplot2 dans SciREPL.

Personnalisez n'importe quel graphique avec les surcharges standard `+ theme(...)` de ggplot2.
