# معرض ggplot2

رسومات بيانية تفاعلية باستخدام ggplot2 مع السمة الداكنة لـ SciREPL — تُطبق تلقائيًا دون الحاجة إلى إعدادات إضافية.

يستخدم مجموعات بيانات R المدمجة (`mpg` و`mtcars`). لا حاجة لتنزيل أي بيانات.

```r
install.packages("ggplot2")
library(ggplot2)
cat("تم تحميل ggplot2 — تم تطبيق theme_scirepl() تلقائيًا\n")
```

## 1. مخطط مبعثر مع خط الاتجاه

```r
ggplot(mpg, aes(displ, hwy, color = class)) +
  geom_point(size = 2, alpha = 0.7) +
  geom_smooth(method = "lm", se = FALSE, color = "#58a6ff") +
  labs(title = "حجم المحرك مقابل استهلاك الوقود على الطريق السريع (MPG)",
       x = "سعة المحرك (لتر)", y = "استهلاك الوقود على الطريق السريع (MPG)")
```

## 2. مخطط شريطي مقسم إلى أوجه (Faceted)

```r
ggplot(mpg, aes(class, fill = drv)) +
  geom_bar(position = "dodge") +
  facet_wrap(~year) +
  labs(title = "فئات المركبات حسب نوع الدفع (1999 مقابل 2008)",
       x = "الفئة", y = "العدد") +
  theme(axis.text.x = element_text(angle = 45, hjust = 1))
```

## 3. مخطط الكثافة

```r
ggplot(mpg, aes(hwy, fill = drv)) +
  geom_density(alpha = 0.5) +
  labs(title = "توزيع استهلاك الوقود على الطريق السريع حسب نوع الدفع",
       x = "استهلاك الوقود على الطريق السريع (MPG)", y = "الكثافة")
```

## 4. مخطط الصندوق مع التشتت النقطي (Jitter)

```r
ggplot(mpg, aes(class, hwy, fill = class)) +
  geom_boxplot(alpha = 0.7, show.legend = FALSE) +
  geom_jitter(width = 0.2, alpha = 0.3, size = 1) +
  labs(title = "استهلاك الوقود على الطريق السريع حسب فئة المركبة",
       x = "الفئة", y = "استهلاك الوقود على الطريق السريع (MPG)") +
  coord_flip()
```

## 5. خريطة الارتباط الحرارية

```r
corr <- round(cor(mtcars), 2)
vars <- colnames(mtcars)
grid <- expand.grid(x = vars, y = vars)
grid$value <- as.vector(corr)
ggplot(grid, aes(x, y, fill = value)) +
  geom_tile(color = "#30363d") +
  geom_text(aes(label = ifelse(abs(value) > 0.5, value, "")),
            color = "#e6edf3", size = 2.5) +
  scale_fill_gradient2(low = "#f85149", mid = "#161b22", high = "#58a6ff",
                       midpoint = 0, limits = c(-1, 1)) +
  labs(title = "خريطة الارتباط الحرارية لـ mtcars", fill = "r") +
  theme(axis.text.x = element_text(angle = 45, hjust = 1))
```

## الملخص

تستخدم جميع المخططات أعلاه `theme_scirepl()`، التي تُطبق تلقائيًا عند تحميل ggplot2 في SciREPL.

يمكنك تخصيص أي مخطط باستخدام تجاوزات `+ theme(...)` القياسية في ggplot2.
